```python
import pandas as pd
import numpy as np

df = pd.read_csv("Slack_queries.csv")

print(df.head())
print(df.info())
print(df.describe(include='all'))
print(df.shape)
```

```python
missing = df.isnull().sum()

missing_percent = (
    df.isnull().sum()/len(df)*100
).round(2)

missing_df = pd.DataFrame({
    "Missing Values":missing,
    "Percentage":missing_percent
})

print(missing_df)
```

```python
status_counts = df["Status (Ticket)"].value_counts()

print(status_counts)

print(
    status_counts/len(df)*100
)
```

```python
user_counts = df["Student or WP"].value_counts()

print(user_counts)

print(
    df["Student or WP"].value_counts(normalize=True)*100
)
```

```python
program_counts = df["Program Name"].value_counts()

print(program_counts)
```

```python
pd.crosstab(
    df["Program Name"],
    df["Status (Ticket)"]
)
```

```python
phase_counts = df["Project Phase"].value_counts()

print(phase_counts)
```

```python
date_cols = [
    "Created Time (Ticket)",
    "Ticket Closed Time",
    "First Response Time"
]

for col in date_cols:
    df[col] = pd.to_datetime(
        df[col],
        errors="coerce",
        dayfirst=True
    )
```

```python
df["Hour"] = df["Created Time (Ticket)"].dt.hour

hourly = df.groupby("Hour").size()

print(hourly)
```
